## Business Context

Telecommunication networks generate massive volumes of time-series data
coming from:

- radio access network (RAN)
- core network
- transport network

Operators must continuously monitor:

- traffic load
- latency
- packet loss
- throughput
- signaling behavior

Failures or congestion may lead to:

- service degradation
- customer dissatisfaction
- SLA violations
- revenue loss

This project aims to build an anomaly detection framework to support
network supervision and operational excellence.

```python
import torch
import torch.nn as nn

class LSTMAutoencoder(nn.Module):
    def __init__(self, n_features, hidden_size=64):
        super().__init__()

        self.encoder = nn.LSTM(
            input_size=n_features,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True
        )

        self.decoder = nn.LSTM(
            input_size=hidden_size,
            hidden_size=n_features,
            num_layers=2,
            batch_first=True
        )

    def forward(self, x):
        encoded, _ = self.encoder(x)
        decoded, _ = self.decoder(encoded)
        return decoded
```

```python
class LSTMAutoencoder(nn.Module):
    def __init__(self, n_features, hidden_size=64):
        super().__init__()

        self.encoder = nn.LSTM(
            input_size=n_features,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True
        )

        self.decoder = nn.LSTM(
            input_size=hidden_size,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True
        )

        self.output_layer = nn.Linear(hidden_size, n_features)

    def forward(self, x):
        encoded, _ = self.encoder(x)
        decoded, _ = self.decoder(encoded)
        reconstructed = self.output_layer(decoded)
        return reconstructed
```

```python
model = LSTMAutoencoder(n_features=6).to(device)
model
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import pandas as pd

data = dataset["data"]

selected_kpis = ["RSRP", "DL_BLER", "UL_BLER", "UL_MCS", "TX_Bytes", "RX_Bytes"]
WINDOW_SIZE = 20

def make_windows(arr, window_size=20):
    return np.array([arr[i:i+window_size] for i in range(len(arr) - window_size)])

normal_windows = []
max_normal_samples = 80  # on démarre petit, on augmentera après

count = 0
for i in range(len(data)):
    if data[i]["labels"]["anomaly_present"] == "No":
        df_i = pd.DataFrame(data[i]["KPIs"])[selected_kpis]

        scaler = MinMaxScaler()
        arr_i = scaler.fit_transform(df_i.values)

        X_i = make_windows(arr_i, WINDOW_SIZE)
        normal_windows.append(X_i)

        count += 1
        if count >= max_normal_samples:
            break

X_train = np.concatenate(normal_windows, axis=0)
X_train.shape
```

```python
import torch
from torch.utils.data import DataLoader, TensorDataset
import torch.nn as nn

device = "cuda" if torch.cuda.is_available() else "cpu"

X_train_t = torch.tensor(X_train, dtype=torch.float32)
train_loader = DataLoader(TensorDataset(X_train_t, X_train_t), batch_size=128, shuffle=True)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

EPOCHS = 8

model.train()
for epoch in range(1, EPOCHS + 1):
    epoch_loss = 0.0
    for xb, yb in train_loader:
        xb = xb.to(device)
        yb = yb.to(device)

        optimizer.zero_grad()
        yhat = model(xb)
        loss = criterion(yhat, yb)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * xb.size(0)

    epoch_loss /= len(train_loader.dataset)
    print(f"Epoch {epoch}/{EPOCHS} - train_loss: {epoch_loss:.6f}")
```

```python
import numpy as np
import torch

model.eval()

def reconstruction_errors(X_np, batch_size=256):
    X_t = torch.tensor(X_np, dtype=torch.float32)
    loader = DataLoader(TensorDataset(X_t, X_t), batch_size=batch_size, shuffle=False)

    errs = []
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)

            yhat = model(xb)
            # MSE per window (mean over time and features)
            e = torch.mean((yhat - yb) ** 2, dim=(1,2))
            errs.append(e.detach().cpu().numpy())

    return np.concatenate(errs)

train_err = reconstruction_errors(X_train)
threshold = np.percentile(train_err, 99)

train_err.min(), train_err.mean(), train_err.max(), threshold
```

```python
# Rebuild the jamming sample window tensor exactly like before
sample = dataset["data"][0]   # (si c'est bien ton jamming)
df_sample = pd.DataFrame(sample["KPIs"])[selected_kpis]

scaler = MinMaxScaler()
arr_sample = scaler.fit_transform(df_sample.values)

X_sample = make_windows(arr_sample, WINDOW_SIZE)
err_sample = reconstruction_errors(X_sample)

# Plot anomaly score
import matplotlib.pyplot as plt

plt.figure(figsize=(15,4))
plt.plot(err_sample)
plt.axhline(threshold, linestyle="--")
plt.title("Anomaly score (reconstruction error) - sample")
plt.xlabel("Window index")
plt.ylabel("MSE")
plt.show()

print("Anomaly present:", sample["labels"]["anomaly_present"])
print("Anomaly type:", sample["anomalies"]["type"])
print("Anomaly duration:", sample["anomalies"]["anomaly_duration"])
print("Percent windows above threshold:", (err_sample > threshold).mean())
```
